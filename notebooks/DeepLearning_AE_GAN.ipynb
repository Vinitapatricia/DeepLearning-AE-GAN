```python
import torch, matplotlib.pyplot as plt
from torchvision.utils import make_grid
from src.config import DEVICE, NOISE_DIM
from src.models.gan import Generator
G = Generator(NOISE_DIM).to(DEVICE)
G.load_state_dict(torch.load("outputs/checkpoints/gan_G.pt")); G.eval()
with torch.no_grad():
    g = (G(torch.randn(16, NOISE_DIM, device=DEVICE)).cpu() + 1) / 2
plt.imshow(make_grid(g, nrow=4).permute(1,2,0)); plt.axis("off"); plt.show()
```

```python
import torch, matplotlib.pyplot as plt, time
from torchvision.utils import make_grid
from src.config import DEVICE
from src.models.ddpm import UNet, Diffusion

model = UNet().to(DEVICE)
model.load_state_dict(torch.load("outputs/checkpoints/ddpm.pt")); model.eval()
diff = Diffusion(1000, DEVICE)
t0 = time.time()
x = diff.sample(model, 16, DEVICE)
print("waktu sampling 16 citra:", time.time() - t0, "detik")
g = (x.cpu() + 1) / 2
plt.imshow(make_grid(g, nrow=4).permute(1, 2, 0)); plt.axis("off"); plt.show()
```
